# Basic ChatBot

A Jupyter notebook based on tutorials on using n-gram techniques to implement simple chatbots on Brilliant.org

```python
# Dependencies
import random
from abc import ABC, abstractmethod
```

```python
class BrainInterface(ABC):
    """Interface for chatbot brains."""

    @abstractmethod
    def process(self, textfile) -> None:
        """Learn from an input file."""

    @abstractmethod
    def generate(self, context: str, noutput: int ) -> str:
        """Return text generated from the input string."""
```

```python
class NGram(BrainInterface):
    """
    A class representing an nGram-based predictive text model.
    Attributes:
        nglen: int 
            The n-gram length + 1,
            e.g. nglen=3 implies that 2 words of context will be used to generate each new word.
        word_map: dictionary
            The map between each nglen-gram and the words which follow it in the input text.
        punctuation: str
            The string containing all the characters which will be removed when 
                                parsing the input text.
                                
    Methods:
    
        process(textfile)
            Populate the word_map using the text contained in the file textfile
        generate(context, noutput)
            Generate noutput words of text based on the words contained in the context string 
        
    """

    punctuation='.;,-“’”:?—‘!()_'
    word_map={}
    nglen=2
    
        
    def set_len(self,nglen) -> None:
        """
        Sets the length of the nGram used in the word map
        Parameters
        ----------
            nglen: int
                    The n-gram length + 1
        """
        self.nglen=nglen
        
    def process(self,textfile) -> None:
        """
        Populate the word_map using the text (converted to lowercase and with the characters in the punctuation string removed) contained in the file textfile.

        Parameters
        ----------
            textfile: file object
                    The file to be parsed.
        """
        window=[]
        for line in textfile:
            for word in line.split():
                clean_word = word.strip(NGram.punctuation).lower()
                window.append(clean_word)
                if len(window) == self.nglen:
                  key = tuple(window[0:self.nglen-1])
                  value = window[self.nglen-1]
                  if key in self.word_map:
                    self.word_map[key].append(value)
                  else:
                    self.word_map[key] = [value]
                  window.pop(0)

    def generate(self,context,noutput) -> str:
        """
        Generate noutput words of text based on the words contained in the context string.
        
        Parameters
        ----------
            context: string
                    The string used to give the initial context (the first nglen words) to generate the next word from the word_map.
            noutput: int
                    The number of words of output to generate.
        """
        response=[]
        words=context.split()[0:self.nglen]
       
        for i in range(noutput):
            words.append("blah")
            response.append(words[0])
            next_words = self.word_map[tuple(list(words)[0:self.nglen-1])]
            words[self.nglen-1] = random.choice(next_words)
            words.pop(0)
        return " ".join(response)
 
```

```python
class Bot:
    """
    A class representing a chatbot.
    Attributes:
        name: string 
            The name of the bot.
        brain: 
            The brain model which can process input text and predictively generate output.
       
    Methods:
    
        introduce()
            Print a greeting.
        respond(context, nwords)
            Return 'nwords' of response to the words in the string 'context'.  
        read(booklist)
            Loop over all the files listed in the list 'booklist', and use each one to train the bot's brain
            
    """

    def __init__(self,name,brain=NGram()):
        self.name=name
        self.brain=brain
        
    def introduce(self) ->None:
        """
        Print a greeting.
        """
        print("Hello, I'm", self.name)
        
    def respond(self,context,nwords=5) -> str:
        """
        Return 'nwords' of response to the words in the string 'context'.
        Parameters
        ----------
            context: str
                    The input string used to generate the response.
            nwords: int
                    The number of words of output to generate.
        """
        return self.brain.generate(context, nwords)
    
    def read(self,booklist) ->None:
        """
        Loop over all the files listed in the list 'booklist', and use each one to train the bot's brain.
        Parameters
        ----------
            booklist: list of str
                    A list of strings, where each string contains the filename of a file to be passed to the bot's brain.
        """
        for book in booklist:
            textfile=open(book)
            self.brain.process(textfile)
```

```python
# n-gram model adapted from examples on brilliant.org

if __name__=="__main__":
    
    
    # The list of filenames corresponding to the text files (books) which the bot will read
    datadir="data/"
    books = [datadir + book for book in ["book1.txt","book2.txt","book3.txt","book4.txt", "book5.txt", "book6.txt", "book7.txt", "book8.txt", "book9.txt", "book10.txt", "book11.txt", "book12.txt", "book13.txt", "book14.txt"] ]        
   
    # The list of random seeds to be iterated over when generating the bot's responses
    seeds=[65,754,34,23,5645]
    
    # The size of the window used in the bot's ngram brain (nwin-1 words of context + 1 word of response)
    nwin=4
    
    # The number of words of output to be generated in each response
    noutput=100
    
    # The start of the bot's sentence
    mywords="I am sure"
        
    print("Number of words of context:", nwin-1)
    
    #Create a brain
    brain=NGram()
    brain.set_len(nwin)
    
    # Give the brain to a bot
    sam=Bot("Sam", brain)
    
    # Teach the bot
    sam.read(books)
    
    # Generate responses from the bot
    for seed in seeds:
        
        print("initial words: ", mywords)
        print("Using seed:", seed)
        random.seed(seed)
        print(sam.respond(mywords.lower(),noutput))
        print()
    
```
